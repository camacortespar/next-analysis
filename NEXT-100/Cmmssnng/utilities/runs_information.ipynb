```python
all_runs = []
# Loop through the entire data structure and flatten it
for isotope, runs_dict in RUNS_INFO:
        for run_number, stats in runs_dict.items():
            all_runs.append({
                "run_number"   : run_number,
                "pressure"     : stats["pressure"],
                "cath_voltage" : stats["cath_voltage"],
                "gate_voltage" : stats["gate_voltage"],
                "duration"     : stats["duration"],
                "OK"           : stats["OK"],
                "LOST"         : stats["LOST"],
                "isotope"      : isotope,
            })

# Dataframe
MASTER_DF = pd.DataFrame(all_runs).set_index("run_number")
```
